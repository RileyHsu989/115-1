```python
h= float(input("請輸入身高（公分）："))
w= float(input("請輸入體重（公斤）："))
h = h / 100
bmi = w / (h ** 2)
if bmi < 18.5:
    result = "過輕"
elif 18.5 <= bmi <= 24:
    result = "正常"
else:
    result = "過重"

print(f"你的 BMI 值為：{bmi:.2f}，判定為：{result}")
```
