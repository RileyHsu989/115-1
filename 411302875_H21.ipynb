```python
import datetime
import traceback
try:
  a=int(input("請輸入被除數："))
  b=int(input("請輸入除數："))
  result=a/b
  print(f"結果：{result}")
except Exception as e:
  with open("log_test.txt","a",encoding="utf-8") as f:
    f.write(f"{datetime.datetime.now()}.發生錯誤={e}\n")
    f.write(traceback.format_exc())
finally:
  print("程式結束")
```
