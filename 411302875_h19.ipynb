```python
backpack = set()
print(f"初始背包: {backpack}")

backpack.add("劍")
backpack.add("盾")
backpack.add("藥水")
print(f"撿到道具後的背包: {backpack}")

backpack.add("藥水")
print(f"再次撿到藥水後的背包: {backpack}")

backpack.remove("盾")
print(f"使用盾後的背包: {backpack}")

dropped_item = backpack.pop()
print(f"掉落的道具: {dropped_item}")

print(f"遊戲結束，最後背包: {backpack}")
```
