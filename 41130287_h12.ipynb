```python
def c_bmi(h,w):
  bmi=w/(h/100)**2
  return bmi
bmi1=c_bmi(170,60)
bmi2=c_bmi(165,55)
bmi3=c_bmi(180,80)
print(f"第一位的bmi是{bmi1:.2f}\n第二位的bmi是{bmi2:.2f}\n第三位的bmi是{bmi3:.2f}\n")
```

```python
def f1(p,q):
  return p*q
def f2(p,q):
  total=f1(p,q)
  return total*0.9
p=int(input("請輸單價:"))
q=int(input("請輸數量:"))
print("原價:",f1(p,q))
print("折扣:",f2(p,q))
```
