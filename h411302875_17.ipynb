```python
x = ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
powers = [10, 30, 50, 80, 100]
for i in range(len(x)):
    print(f"{i + 1} {x[i]}")
while True:
  code = int(input("請輸入武器編號（0 結束）："))
  if code == 0:
    break
  elif 1 <= code <= len(x):
    index = code - 1
    print(f"{x[index]} 的攻擊力是 {powers[index]}")
  else:
    print("找不到武器")
avg_power = sum(powers) / len(powers)
print(f"平均攻擊力： {avg_power}")
```
