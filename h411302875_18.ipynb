```python
p = {"蘋果": 50, "香蕉": 30, "牛奶": 60, "麵包": 40}
print("=== 超市商品 ===")
for name, price in p.items():
    print(f"{name}：{price}元")
total_amount = 0
while True:
  item = input("請輸入商品名稱（輸入 / 結束）：")
  if item == "/":
    break
  if item in p:
    total_amount += p[item]
  else:
    print("找不到此商品，請重新輸入！")
print(f"總金額：{total_amount} 元")
```
